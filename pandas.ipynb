```python
import pandas as pd
data=[1,2,3,4,5]
#converting list to series
#indices will be the default indices and values will be the values of the list
series=pd.Series(data)
print(series)
print(type(series))
```

```python
#converting Dictionary to series
#indices will be the keys of the dictionary and values will be the values of the dictionary
data={'a':1,'b':2,'c':3}
series=pd.Series(data)
print(series)
```

```python
data=[10,20,30,40,50]
index1=['a','b','c','d','e']
index2=[1,2,3,4,5]
#converting list to series with custom indices
series1=pd.Series(data,index=index2)
series2=pd.Series(data,index=index1)
print(series1)
print(series2)
```

```python
data={'Name': ['Alice', 'Bob', 'Charlie', 'David'],
      'Age': [25, 30, 35, 40],
        'City': ['New York', 'Los Angeles', 'Chicago', 'Houston']}
df=pd.DataFrame(data)
df
```

```python
#Adding a new column to the DataFrame
df['Country'] = ['USA', 'USA', 'USA', 'USA']
df['Salary'] = [50000, 60000, 70000, 80000]
df
```

```python
#Deleting a column Temperorily from the DataFrame
df.drop('Salary', axis=1) #to delete a column from the DataFrame using drop function
```

```python
df
```

```python
#Deleting a column from the DataFrame permanently
df.drop('Salary', axis=1, inplace=True) #to delete a column from the DataFrame permanently using drop function with inplace parameter set to True
```

```python
df
```

```python
df.drop('Country', axis=1, inplace=False) #to delete a column from the DataFrame permanently using drop function with inplace parameter set to False
```

```python
df
#we can delete a column temperorily by both not using inplace parameter or by setting inplace parameter to False. But if we want to delete a column permanently we have to set inplace parameter to True.
```

```python
import pandas as pd
df=pd.read_csv('data.csv')
df.head()
```

```python
df.describe()
#Statistical summary of the DataFrame using describe function
```

```python
#isnull() function is used to check for missing values in the DataFrame. It returns a DataFrame of the same shape as the original DataFrame, with True indicating missing values and False indicating non-missing values.
df.isnull().head()
```

```python
#any() function is used to check if any of the values in the DataFrame are missing. It returns a boolean value indicating whether any of the values are missing or not. if atleast one value is missing in the DataFrame it will return True otherwise it will return False.
df.isnull().any()
```

```python
#sum() function is used to count the number of missing values in each column of the DataFrame. It returns a Series with the column names as the index and the number of missing values as the values.
df.isnull().sum()
```

```python
#replaces all null values with the specified given value. here:0
df1=df.fillna(0)
df1.head(10)
```

```python
Value_mean=df1.Sales.mean()
Value_mean
#to calculate the mean of the Sales column in the DataFrame df1 using the mean() function. The result is stored in the variable Value_mean and then printed to the console.
```

```python
#filling null values with the mean of the 'Value' column from df1
df['Values_fillNA']=df['Value'].fillna(df1.Value.mean())
df.head(16)
```

```python
#Renaming a column in the DataFrame using rename function
df.rename(columns={'Values_fillNA':'Value_filled_with_mean'})
df
```

```python
#Changing Datatype
from numpy import int64


df['new_Values']=df['Value'].fillna(df['Value'].mean()).astype(int64)
df.drop('new_Values', axis=1, inplace=True)
df
```

```python
df['square_values']=df['Value']**2
#or df['square_values']=df['Value'].apply(lambda x: x**2) 
#to create a new column 'square_values' in the DataFrame df that contains the square of the values in the 'Value' column using apply function with lambda function. The result is stored in the new column 'square_values' and then printed to the console.
df.head()
```
